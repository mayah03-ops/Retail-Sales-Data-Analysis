```python
import pandas as pd

df = pd.read_csv("data/superstore.csv")

print("Dataset loaded successfully!")
print(df.shape)
```

```python
import matplotlib.pyplot as plt
```

```python
sales_by_category = df.groupby("Category")["Sales"].sum()

sales_by_category.plot(kind="bar")

plt.title("Sales by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
profit_by_category = df.groupby("Category")["Profit"].sum()

profit_by_category.plot(kind="bar")

plt.title("Profit by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
profit_by_subcategory = df.groupby("Sub_Category")["Profit"].sum().sort_values()

profit_by_subcategory.plot(kind="bar", figsize=(12, 6))

plt.title("Profit by Product Subcategory")
plt.xlabel("Subcategory")
plt.ylabel("Total Profit")
plt.xticks(rotation=45, ha="right")
plt.axhline(y=0, linestyle="--")
plt.tight_layout()
plt.show()
```

```python
profit_by_region = df.groupby("Region")["Profit"].sum().sort_values(ascending=False)

profit_by_region.plot(kind="bar")

plt.title("Profit by Region")
plt.xlabel("Region")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

profit_by_region.plot(kind="bar")

plt.title("Profit by Region")
plt.xlabel("Region")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)

plt.tight_layout()
plt.savefig("profit_by_region.png", dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sales_by_category.plot(kind="bar")

plt.title("Sales by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)

plt.tight_layout()
plt.savefig("sales_by_category.png", dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

profit_by_category.plot(kind="bar")

plt.title("Profit by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)

plt.tight_layout()
plt.savefig("profit_by_category.png", dpi=300)
plt.show()
```
